# ball battle — match renderer

Renders one headless-browser match to an mp4 you can download.
Works on the **free CPU runtime** (Runtime → Change runtime type → CPU — the GPU is useless for 2D canvas).

Defaults: 720p30, one match ≈ 620 s of content ≈ **~11 min wall time**.

```
# setup — installs system deps + Chromium (Ubuntu's own chromium is a snap, won't run here)
!apt-get update -qq
!apt-get install -y -qq xvfb openbox xdotool pulseaudio ffmpeg fonts-dejavu-core fonts-jetbrains-mono > /dev/null
!pip -q install playwright
!python -m playwright install chromium --with-deps
!ln -sf $(ls /root/.cache/ms-playwright/chromium-*/chrome-linux/chrome | head -1) /usr/local/bin/chromium
!chromium --version && rm -rf ball-battle && git clone https://github.com/LakshanDS/ball-battle
```

```
# render — tune RES / FPS / BITRATE / DURATION here (1080p30 drops frames on 2 free cores)
%%bash
pkill -f 'Xvfb|openbox|chromium' 2>/dev/null; sleep 1
export RES=1920x1080 FPS=60 BITRATE=6000k DURATION=620
export OUTFILE=/content/ball-battle-match.mp4
cd ball-battle/stream && bash browser-stream.sh
ls -lh "$OUTFILE"
```
